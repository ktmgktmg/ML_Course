```python
#%% Data imputation
import numpy as np
from sklearn.impute import SimpleImputer
import scipy.sparse as sp
import pandas as pd
```

```python
imp = SimpleImputer(missing_values=np.nan, strategy = 'most_frequent') #fill_value=0) #mean, median, constant
imp.fit([[1, 2], 
         [np.nan, 3], 
         [1, 2], 
         [7, 6]])
print(imp.statistics_)
```

```python

X = np.array([[np.nan, 2], 
              [6, np.nan], 
              [7, 6]])
print(X)
X=imp.transform(X)
print(X)
```

```python
Y = np.array([[np.nan, 2], [np.nan, np.nan], [7, 6], [7, np.nan]])
print(Y)
```

```python
Y=imp.transform(Y)
print(Y)
```

```python

#%% scattered matrix
X = sp.csc_matrix([[-1, -1], [0, -1], [8, 4]])
print(X)
```

```python
imp2 = SimpleImputer(missing_values=-1, strategy='mean')
imp2.fit(X)
print(X)
print(imp2.statistics_)
```

```python
X_test = sp.csc_matrix([[-1, 2], [6, -1], [7, 6]])
print(X_test)
print(imp2.transform(X_test).toarray())
```

```python
#%% most_frequent
df = pd.DataFrame([["a", "x"],
                   [np.nan, "y"],
                   ["a", np.nan],
                   ["b", "y"]], dtype="category")
df.head()
```

```python
imp3 = SimpleImputer(strategy="most_frequent")
df2 = imp3.fit_transform(df)
print(df2)
print(type(df2))
```

```python
#%%
A = np.array([["Y","N","Y","N"],
              ["Y","N","?","N"],
              ["Y","N","Y","?"],
              ["?","N","Y","Y"]])
df = pd.DataFrame(A,columns=["Airconditioning","Garage","Pool","Verandah"])
df.head()
```

```python

imp4 = SimpleImputer(missing_values="?", strategy='most_frequent')
df2 = pd.DataFrame(imp4.fit_transform(df))
df2.head()
```

```python
#%%
A = np.array([["Y","N","Y","N"],
              ["Y","N","NA","N"],
              ["Y","N","Y","Y"],
              ["NA","N","Y","Y"]])
print(A)
df = pd.DataFrame(A,columns=["Airconditioning","Garage","Pool","Verandah"])
print(df.head())
```

```python

imp = SimpleImputer(missing_values="NA", strategy='most_frequent')
df2 = pd.DataFrame(imp.fit_transform(df))
df2.head()
```

```python

#%% K-mean Algorithm
from sklearn.impute import KNNImputer #KNNImpute machine learning based K-mean algorithm
nan = np.nan
X = np.array([[1, 0, nan], 
              [3, 5, 1], 
              [nan, 6, 5], 
              [8, 8, nan]])
print(X)
```

```python

imp = KNNImputer(n_neighbors=1, weights="uniform")
X=imp.fit_transform(X)
print(X)
```
