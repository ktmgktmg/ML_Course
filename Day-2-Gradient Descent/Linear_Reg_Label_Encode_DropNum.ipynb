```python
# Define a linear regression model
class LinearRegressionModel(tf.keras.Model):
    def __init__(self):
        super(LinearRegressionModel, self).__init__()
        self.dense = tf.keras.layers.Dense(1, input_shape=(1,))

    def call(self, inputs):
        return self.dense(inputs)
```

```python
# Create an instance of the model
model = LinearRegressionModel()

# Define the loss function (Mean Squared Error)
loss_object = tf.keras.losses.MeanSquaredError()

# Define the optimizer (Gradient Descent)
optimizer = tf.keras.optimizers.SGD(learning_rate=0.01)
```

```python
# Define a function to compute the loss and gradients
def train_step(inputs, labels):
    with tf.GradientTape() as tape:
        predictions = model(inputs)
        loss = loss_object(labels, predictions)
    gradients = tape.gradient(loss, model.trainable_variables)
    optimizer.apply_gradients(zip(gradients, model.trainable_variables))
    return loss
```

```python
# Train the model
num_epochs = 1000
for epoch in range(num_epochs):
    loss_value = train_step(X_train_tensor, y_train_tensor)
    if epoch % 100 == 0:
        print("Epoch:", epoch, "Loss:", loss_value.numpy())

# Get the final weights and bias
final_weights, final_bias = model.get_weights()

# Print the final weights and bias
print("Final Weights:", final_weights)
print("Final Bias:", final_bias)
```

```python
predictions = model(X_train_tensor)

# Calculate R-squared value
y_mean = tf.reduce_mean(y_train_tensor)
TSS = tf.reduce_sum(tf.square(y_train_tensor - predictions))
RSS = tf.reduce_sum(tf.square(y_train_tensor - y_mean))
R_squared = 1 - (RSS / TSS)

# Print R-squared value
print("R-squared value:", R_squared.numpy())
```

```python
from sklearn.decomposition import PCA
```

```python
# Initialize PCA
pca = PCA()
```

```python
# Fit PCA to the data
pca.fit(X_N)
```

```python
# Explained variance ratio
eigVal = pca.explained_variance_ratio_
```

```python
# Cumulative explained variance ratio
sumeigVal = np.sum(eigVal)
```

```python
# Number of components to select
k = 55  # Example: selecting the first 2 principal components

# Sum of selected eigenvalues
sumSel = np.sum(eigVal[:k])
# Calculate the ratio
ratio = sumSel / sumeigVal

print(f"Ratio of selected eigenvalues to total eigenvalues: {ratio:.2f}")
```

```python
# Initialize PCA with the number of components to retain
pca_reduced = PCA(n_components=k)
# Fit and transform the data to the new reduced dimensionality
X_reduced = pca_reduced.fit_transform(X_N)
X_reduced.shape
```
