```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = np.array([[200, 2, 300],
              [300, 1, 500],
              [150, 0, 200]])
```

```python
dataset = pd.DataFrame(data,columns =['size','garage car', 'price'])
dataset
```

```python
m = dataset.shape[0]
m
```

```python
X1 = dataset[['size','garage car']].values
y = dataset['price'].values
X1, y
```

```python
X = np.c_[np.ones((3,1)), X1]
X
```

```python
theta = np.array([[0], [2], [2]])
theta
```

```python
h = X@theta
h,y
```

```python
# Create a new figure
plt.figure(figsize=(8, 5))

# Plot the data
plt.plot(X[:,1], y, 'x',label='Target',  color='black')
plt.plot(X[:,1], h, 'o',label='Predicted', color='blue')
# Add title and axis labels
plt.title('Line Plot with Tight Axis')
plt.xlabel('Size [ft^2]')
plt.ylabel('price [1000 $]')

# Tight axis
plt.axis('tight')

# Show legend and grid
plt.legend()
plt.grid(True)

# Display the plot
plt.show()
```

```python
J = 1/(2*m)*np.sum((h-y)**2)
J
```
