```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
m = 10
x = np.arange(0,m+1) 
y = 2*x + 2 + np.random.rand(m+1) - np.random.rand(m+1)
x,y
```

```python
plt.plot(x,y,'b-o')
```

```python
def cost2(X,y,theta):
    m = X.shape[0]
    J = 1/(2*m)*np.sum((X@theta-y)**2)
    dJdtheta = 1/m*(X.T@(X@theta-y))
    return J, dJdtheta
```

```python
theta = np.array([0, -1])
theta
```

```python
m = x.shape[0]
X = np.c_[(np.ones((m,1)),x.reshape(-1,1))]
X[:5,:]
```

```python
J,dJdtheta = cost2(X,y,theta)
J, dJdtheta
```

```python
theta = np.array([0, -1.0])
alpha = 0.05
maxIter = 500
J = np.zeros((maxIter,1))
for i in range(maxIter):
    J[i],dJdtheta = cost2(X,y,theta)
    theta -= alpha*dJdtheta
    if i%100 == 0:
        print(f"{i:d}, J = {J[i][0]:.2f}")
```

```python
plt.plot(np.arange(maxIter),J,'b--')
plt.xlabel('iteration')
plt.ylabel('cost function')
```

```python
theta
```

```python
h = X@theta
h[:5],y[:5]
```

```python
SS_tot = np.sum((y-np.mean(y))**2)
SS_res = np.sum((y-h)**2)
R_squared = 1- SS_res/SS_tot
R_squared,SS_res
```

```python
plt.plot(x,y,'b-o')
plt.plot(x,h,'r--+')
plt.xlabel('x')
plt.ylabel('y')
plt.legend(['actual','predict'])
```
