## Derive the equation for gradient functions for the cost function given below.
$$
\text{Hypothesis: } h_\theta(x) = \theta_0 + \theta_1 x
$$

$$
\text{Cost function: } 
J(\theta_0, \theta_1) = \frac{1}{2m} \sum_{i=1}^{m} \left( h_\theta(x^{(i)}) - y^{(i)} \right)^2
$$

```python
# import required libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Generate data samples for fitting
m = 10
x = np.arange(0,m+1) 
y = 2*x + 2 + np.random.rand(m+1) - np.random.rand(m+1)
x,y
```

```python
# visualize the data
plt.plot(x,y,'b-o')
```

```python
# This the Python function to compute the cost and gradient functions.
# Insert your code for the formula for computing the cost function. The gradient function is already coded for you.
def cost2(X,y,theta):
    m = X.shape[0]
    # Inser code for cost function here. 
    dJdtheta = 1/m*(X.T@(X@theta-y))
    return J, dJdtheta
```

```python
# initializing theta 
theta = np.array([0, -1])
theta
```

```python
# prepare dataset for training model
m = x.shape[0]
X = np.c_[(np.ones((m,1)),x.reshape(-1,1))]
X[:5,:]
```

```python
# Test the cost function here by calling the function you have just defined. If it works, it should generate
# the cost function and its derivative.
J,dJdtheta = cost2(X,y,theta)
J, dJdtheta
```

```python
# train the model to obtain optimal values of theta_0 and theta_1 for minimize=ing J
theta = np.array([0, -1.0])
alpha = 0.05
maxIter = 500
J = np.zeros((maxIter,1))
for i in range(maxIter):
    J[i],dJdtheta = cost2(X,y,theta)
    theta -= alpha*dJdtheta
    if i%100 == 0:
        print(f"{i:d}, J = {J[i][0]:.2f}")
```

```python
# plot the convergence
plt.plot(np.arange(maxIter),J,'b--')
plt.xlabel('iteration')
plt.ylabel('cost function')
```

```python
theta
```

```python
# checking the hypothesis function
h = X@theta
h[:5],y[:5]
```

```python
# compute errors and evalute quality of fit
SS_tot = np.sum((y-np.mean(y))**2)
SS_res = np.sum((y-h)**2)
R_squared = 1- SS_res/SS_tot
R_squared,SS_res
```

```python
# compare hypothesis with dataset
plt.plot(x,y,'b-o')
plt.plot(x,h,'r--+')
plt.xlabel('x')
plt.ylabel('y')
plt.legend(['actual','predict'])
```

```python
# predict the target at desired x
y_predict = X_predict@theta
y_predict
```
